```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
```

```python
df = pd.read_csv("/content/Tweets.csv")
df = df[["text", "airline_sentiment"]].copy()
df.columns = ["text", "label"]
print("Original shape:", df.shape)
print(df.head())
print("\nMissing values before cleaning:\n", df.isnull().sum())
print("\nClass distribution:\n", df["label"].value_counts())
df.info()
```

```python

print("\nSentiment Percentages:")
print(df["label"].value_counts(normalize=True) * 100)

print("\nObservation:")
print(f"The dataset contains {len(df)} rows.")
print("The text column contains tweet messages.")
print("The label column contains sentiment categories.")
print("The class distribution indicates whether the")
print("sentiment categories are balanced.")


# ==========================================
# 2. DATA CLEANING
# ==========================================

print("\nMissing Values Before Cleaning:")
print(df[["text", "label"]].isnull().sum())

df = df.dropna(subset=["text", "label"]).copy()

print("\nMissing Values After Cleaning:")
print(df[["text", "label"]].isnull().sum())

duplicates_before = df.duplicated(
    subset=["text", "label"]
).sum()

print("\nDuplicates Before Removal:", duplicates_before)

df = df.drop_duplicates(
    subset=["text", "label"]
).reset_index(drop=True)

print("Duplicates After Removal:",
      df.duplicated(subset=["text", "label"]).sum())


# ==========================================
# 3. FEATURE ENGINEERING
# ==========================================

# Character length may capture differences in tweet detail
df["text_length"] = df["text"].str.len()

# Word count may distinguish short and detailed tweets
df["word_count"] = df["text"].str.split().str.len()

# Exclamation marks may signal emotional emphasis
df["exclamation_count"] = df["text"].str.count("!")

# Mentions may indicate customer service interactions
df["mention_count"] = df["text"].str.count("@")

print("\nEngineered Features:")
print(df[
    ["text_length", "word_count",
     "exclamation_count", "mention_count"]
].describe())


# ==========================================
# 4. OUTLIER ANALYSIS
# ==========================================

Q1 = df["text_length"].quantile(0.25)
Q3 = df["text_length"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["text_length"] < lower) |
    (df["text_length"] > upper)
]

print("\nLength Outliers:", len(outliers))
print("IQR Upper Bound:", upper)

print(
    "Decision: Extreme tweet lengths are retained "
    "because longer or shorter tweets can be valid "
    "and may contain useful sentiment information."
)


# ==========================================
# 5. EDA CHART 1: SENTIMENT DISTRIBUTION
# ==========================================

plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    x="label",
    order=["negative", "neutral", "positive"]
)
plt.title("Distribution of Tweet Sentiment")
plt.xlabel("Sentiment")
plt.ylabel("Number of Tweets")
plt.tight_layout()
plt.show()


# ==========================================
# 6. EDA CHART 2: TWEET LENGTH HISTOGRAM
# ==========================================

plt.figure(figsize=(8, 5))
sns.histplot(
    data=df,
    x="text_length",
    bins=30,
    kde=True
)
plt.title("Distribution of Tweet Length")
plt.xlabel("Characters per Tweet")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()


# ==========================================
# 7. EDA CHART 3: LENGTH BY SENTIMENT
# ==========================================

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="label",
    y="text_length",
    order=["negative", "neutral", "positive"]
)
plt.title("Tweet Length by Sentiment")
plt.xlabel("Sentiment")
plt.ylabel("Tweet Length")
plt.tight_layout()
plt.show()


# ==========================================
# 8. EDA CHART 4: WORD COUNT COMPARISON
# ==========================================

plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x="label",
    y="word_count",
    order=["negative", "neutral", "positive"],
    estimator=np.mean,
    errorbar=None
)
plt.title("Average Word Count by Sentiment")
plt.xlabel("Sentiment")
plt.ylabel("Average Word Count")
plt.tight_layout()
plt.show()


# ==========================================
# 9. EDA CHART 5: CORRELATION HEATMAP
# ==========================================

eda_df = df.copy()

label_map = {
    "negative": 0,
    "neutral": 1,
    "positive": 2
}

eda_df["sentiment_encoded"] = eda_df["label"].map(label_map)

corr = eda_df[
    ["text_length", "word_count",
     "exclamation_count", "mention_count",
     "sentiment_encoded"]
].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()


# ==========================================
# 10. FIVE DATASET-BASED INSIGHTS
# ==========================================

counts = df["label"].value_counts()
percentages = df["label"].value_counts(normalize=True) * 100

length_means = df.groupby("label")["text_length"].mean()
word_means = df.groupby("label")["word_count"].mean()
exclamation_means = (
    df.groupby("label")["exclamation_count"].mean()
)

print("\n========== FIVE KEY DATASET INSIGHTS ==========")

print(
    f"1. The dataset contains {len(df)} cleaned tweets. "
    f"Negative: {percentages.get('negative', 0):.2f}%, "
    f"Neutral: {percentages.get('neutral', 0):.2f}%, "
    f"Positive: {percentages.get('positive', 0):.2f}%."
)

print(
    f"2. {duplicates_before} duplicate tweet-label "
    f"pairs were removed during cleaning."
)

print(
    f"3. The IQR method identified {len(outliers)} "
    f"unusual tweet lengths. These were retained."
)

for sentiment in ["negative", "neutral", "positive"]:
    print(
        f"   {sentiment}: "
        f"{length_means.get(sentiment, 0):.1f} "
        f"characters on average."
    )

print(
    "4. Average word counts by sentiment: " +
    ", ".join(
        f"{s}={word_means.get(s, 0):.1f}"
        for s in ["negative", "neutral", "positive"]
    )
)

print(
    "5. Average exclamation marks per tweet: " +
    ", ".join(
        f"{s}={exclamation_means.get(s, 0):.2f}"
        for s in ["negative", "neutral", "positive"]
    )
)
```

```python
df = df.dropna(subset=["text", "label"]).copy()
print("Duplicate text-label pairs:", df.duplicated(subset=["text", "label"]).sum())
df = df.drop_duplicates(subset=["text", "label"]).reset_index(drop=True)
print("Missing values after cleaning:\n", df.isnull().sum())
print("Duplicates after cleaning:", df.duplicated(subset=["text", "label"]).sum())

class_names = ["negative", "neutral", "positive"]
label_map = {name: idx for idx, name in enumerate(class_names)}
df["label_encoded"] = df["label"].map(label_map)
assert df["label_encoded"].notna().all()
print("Label mapping:", label_map)
print("Cleaned shape:", df.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df["text"].astype(str),
    df["label_encoded"].astype(int),
    test_size=0.2,
    random_state=42
)
```

```python
tokenizer = Tokenizer(num_words=5000)
tokenizer.fit_on_texts(X_train)
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq = tokenizer.texts_to_sequences(X_test)
```

```python
MAX_LEN=50
X_train_pad = keras.preprocessing.sequence.pad_sequences(X_train_seq, maxlen=MAX_LEN,padding="post",truncating="post")
X_test_pad = keras.preprocessing.sequence.pad_sequences(X_test_seq, maxlen=MAX_LEN,padding="post",truncating="post")
print("Train shape:", X_train_pad.shape)
print("Test shape:", X_test_pad.shape)
```

```python
lstm_model=keras.Sequential([layers.Input(shape=(MAX_LEN,)),
                             layers.Embedding(input_dim=5000,output_dim=32),
                             layers.LSTM(32),
                             layers.Dense(3, activation="softmax")
])
lstm_model.summary()
```

```python
lstm_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
history=lstm_model.fit(X_train_pad,y_train,epochs=5,batch_size=35,validation_split=0.2,verbose=1)
```

```python
loss,accuracy=lstm_model.evaluate(X_test_pad,y_test)
```

```python
plt.plot(history.history['accuracy'],label='training')
plt.plot(history.history['val_accuracy'],label='validation')
```

```python
plt.plot(history.history['loss'],label='training')
plt.plot(history.history['val_loss'],label='validation')
```

```python
gru_model=keras.Sequential([layers.Input(shape=(MAX_LEN,)),
                            layers.Embedding(input_dim=5000,output_dim=32),
                            layers.GRU(32),
                            layers.Dense(3,activation='softmax')])
gru_model.summary()
```

```python
gru_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
history=gru_model.fit(X_train_pad,y_train,epochs=5,batch_size=35,validation_split=0.2,verbose=1)
```

```python
loss,accuracy=gru_model.evaluate(X_test_pad,y_test)
```

```python
plt.plot(history.history['accuracy'],label='training')
plt.plot(history.history['val_accuracy'],label='validation')
```

```python
plt.plot(history.history['loss'],label='training')
plt.plot(history.history['val_loss'],label='validation')
```

```python
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("LSTM Training vs Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("LSTM Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()
```
